```python
import numpy as np
temperatures_w1 = np.array([22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9])
temperatures_w1_f = (temperatures_w1 * 9/5) + 32
print("Temperatures in Fahrenheit:", temperatures_w1_f)
max_temp = np.max(temperatures_w1)  
min_temp = np.min(temperatures_w1)  
mean_temp = np.mean(temperatures_w1) 

print(f"Maximum Temperature (Celsius): {max_temp}°C")
print(f"Minimum Temperature (Celsius): {min_temp}°C")
print(f"Mean Temperature (Celsius): {mean_temp:.2f}°C")
```

```python
marks = pd.Series([95, 92, 89, 85, 80], 
index=['Rank1', 'Rank2', 'Rank3', 'Rank4', 'Rank5'])
rank1_mark = marks.iloc[0]
print("Mark of 1st rank student:", rank1_mark)
top_3_loc = marks.loc[['Rank1', 'Rank2', 'Rank3']]
print("\nTop 3 ranks using loc:\n", top_3_loc)
rank3_iloc = marks.iloc[2]
print("\nMark of 3rd rank student using iloc:", rank3_iloc)
above_90 = marks[marks > 90]
print("\nRanks with marks > 90:\n", above_90)
```

```python
marks = pd.Series([95, 92, 89, 85, 80], 
index=['Rank1', 'Rank2', 'Rank3', 'Rank4', 'Rank5'])
marks.loc['Rank1'] = 100
print("After modifying 1st rank mark to 100:\n", marks)
marks = marks.drop('Rank5')
print("\nAfter removing last rank ('Rank5'):\n", marks)
cgpa = marks / 10
print("\nCalculated CGPA:\n", cgpa)
```

```python
data = {
    'TransactionID': [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    'ProductCategory': [
        'Electronics', 'Clothing', 'Electronics', 'Furniture',
        'Clothing', 'Electronics', 'Furniture', 'Clothing',
        'Furniture', 'Electronics'
    ],
    'Region': [
        'North', 'South', 'North', 'East', 'West',
        'North', 'East', 'West', 'South', 'North'
    ],
    'Amount': [200, 150, 300, 450, 200, 250, 300, 180, 350, 400]
}

transactions = pd.DataFrame(data)
print("Original DataFrame:")
print(transactions)

print("\nFirst 5 Rows:")
print(transactions.head())

print("\nLast 5 Rows:")
print(transactions.tail())

print("\nShape:")
print(transactions.shape)

print("\nColumn Names:")
print(transactions.columns)

print("\nData Types:")
print(transactions.dtypes)

print("\nBasic Information:")
transactions.info()

print("\nProductCategory and Amount:")
print(transactions[['ProductCategory', 'Amount']])

print("\nLast 3 Columns:")
print(transactions.iloc[:, -3:])

print("\nNorth Region with Amount > 200:")
print(transactions[
    (transactions['Region'] == 'North') &
    (transactions['Amount'] > 200)
])
print("\nProduct Category Value Counts:")
print(transactions['ProductCategory'].value_counts())
print("\nUnique Regions:")
print(transactions['Region'].unique())
print("\nMean Amount by Region:")
print(transactions.groupby('Region')['Amount'].mean())
transactions.loc[
    transactions['TransactionID'] == 102, 'Amount'
] = 165
print("\nAfter Modifying Amount:")
print(transactions)
transactions['Discount'] = transactions['Amount'] * 0.10
print("\nAfter Adding Discount:")
print(transactions)
transactions = transactions[
    transactions['TransactionID'] != 109
]
print("\nAfter Removing TransactionID 109:")
print(transactions)
transactions =transactions.drop(columns=['discount'], errors='ignore')
print("\nFinal DataFrame:")
print(transactions)
```
